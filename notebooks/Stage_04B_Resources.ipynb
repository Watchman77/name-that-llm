```python
# Stage 04B resource preparation only: no fitting or competition submission.
from pathlib import Path
from datetime import datetime, timezone
import os,sys,json,hashlib,subprocess,base64,gzip,platform
os.environ['HF_HUB_DISABLE_XET']='1'
from huggingface_hub import snapshot_download
ROOT=Path('/kaggle/working')
MODEL_ID='microsoft/deberta-v3-base'
REVISION='8ccc9b6f36199bec6961081d44eb72fb3f7353f3'
REFERENCE_SHA='15ca5d9d95fead2690a6c81559a3862f12143f83ad65684d1c584a2827e92ab3'
MODEL_DIR=ROOT/'deberta-v3-base'
WHEELS=ROOT/'wheels'
WHEELS.mkdir(exist_ok=True)
SNAPSHOT=snapshot_download(MODEL_ID,revision=REVISION,local_dir=MODEL_DIR,token=False,
    allow_patterns=['config.json','pytorch_model.bin','spm.model','tokenizer_config.json','README.md'])
subprocess.run([sys.executable,'-m','pip','download','--only-binary=:all:','--no-deps',
    '--dest',str(WHEELS),'transformers==4.56.2','huggingface-hub==0.36.2',
    'tokenizers==0.22.2','sentencepiece==0.2.1','protobuf==6.31.1'],check=True)
REFERENCE_B64=''
ref=gzip.decompress(base64.b64decode(REFERENCE_B64))
assert hashlib.sha256(ref).hexdigest()==REFERENCE_SHA
(ROOT/'stage03_oof_predictions.csv').write_bytes(ref)
manifest={'status':'resources_ready','model_id':MODEL_ID,'revision':REVISION,'licence':'MIT',
    'reference_sha256':REFERENCE_SHA,'prepared_utc':datetime.now(timezone.utc).isoformat(),
    'preparation_python':platform.python_version(),
    'model_files_sha256':{str(p.relative_to(MODEL_DIR)):hashlib.sha256(p.read_bytes()).hexdigest()
        for p in sorted(MODEL_DIR.rglob('*')) if p.is_file() and '.cache' not in p.parts},
    'wheel_files_sha256':{p.name:hashlib.sha256(p.read_bytes()).hexdigest() for p in sorted(WHEELS.glob('*.whl'))}}
(ROOT/'resources_manifest.json').write_text(json.dumps(manifest,indent=2))
print(json.dumps(manifest,indent=2))
print('Ready for separate offline training. No classifier was fitted.')
```
